```python
import pandas as pd
```

```python
df = pd.read_json("books/dsm-tree/dsm5_selected_chapters_tree.jsonl", lines=True)

df
```

```python
first_paragraphs_df = df.groupby(['book_name','page']).nth(1).sort_values("page")
first_paragraphs_df
```

```python
last_paragraphs_df = df.groupby(['book_name','page']).nth(-1).sort_values("page")
last_paragraphs_df
```

```python
last_paragraphs_df = last_paragraphs_df[last_paragraphs_df['node_type_in_tree']=='leaf']
last_paragraphs_df
```

```python
last_paragraphs_df['next_page'] = None
last_paragraphs_df['next_page_type'] = None
next_pages = []
next_pages_type = []

for i, row in last_paragraphs_df.iterrows():
    _page = row['page']
    book_name = row['book_name']
    wanted_page = _page + 1
    next_page = first_paragraphs_df[(first_paragraphs_df['book_name']==book_name) & (first_paragraphs_df['page']==wanted_page)][['content', "node_type_in_tree"]]
    if(not next_page['content'].values):
        next_pages.append(None)
        next_pages_type.append(None)
    else:
        next_pages.append(next_page['content'].values[0])
        next_pages_type.append(next_page['node_type_in_tree'].values[0])

next_pages, next_pages_type
```

```python
last_paragraphs_df['next_page'] = next_pages
last_paragraphs_df['next_page_type'] = next_pages_type
```

```python
last_paragraphs_df[last_paragraphs_df['next_page'].isna()]
```

```python
last_paragraphs_df =last_paragraphs_df[last_paragraphs_df['next_page'].notna()]
last_paragraphs_df
```

```python
last_paragraphs_df = last_paragraphs_df[last_paragraphs_df['next_page_type']=='leaf']
last_paragraphs_df
```

```python
last_paragraphs_df['content'] = last_paragraphs_df['content'] +  " " + last_paragraphs_df['next_page']
last_paragraphs_df
```

```python
last_paragraphs_df['node_id'] = "with_next_page_" + last_paragraphs_df['node_id']
last_paragraphs_df
```

```python
last_paragraphs_df.to_json("books/dsm-tree/last_concated_with_first_nodes.jsonl", lines=True, orient='records')
```
